This starter notebook loads `properties_modeling_v4.csv` and shows a summmary of the damages.

If you're in Google Colab, be sure to upload `properties_modeling_v4.csv` to your current session.


```python
from pathlib import Path

import pandas as pd

# Colab: after uploading, e.g. DATA_PATH = Path("/content/properties_modeling_v4.csv")
DATA_PATH = Path("properties_modeling_v4.csv")

if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Put properties_modeling_v4.csv at {DATA_PATH.resolve()} or update DATA_PATH"
    )

df = pd.read_csv(DATA_PATH, low_memory=False)
print(f"rows = {len(df):,}  columns = {df.shape[1]}")
df.head()
```

```python
# Damage counts by type (0 = No Damage … 4 = Destroyed)
print(df["damage_ordinal"].value_counts().sort_index())
```
